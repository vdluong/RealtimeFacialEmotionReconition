```python
import pandas as pd
import numpy as np
from tensorflow.keras.utils import to_categorical

print("Đang đọc file dữ liệu, có thể mất vài giây...")
df = pd.read_csv('fer2013.csv')


X_train, y_train, X_test, y_test = [], [], [], []

for index, row in df.iterrows():
    # Chuyển chuỗi pixel thành ma trận 48x48
    pixels = np.array(row['pixels'].split(' '), dtype='float32').reshape(48, 48)
    if row['Usage'] == 'Training':
        X_train.append(pixels)
        y_train.append(row['emotion'])
    else:
        X_test.append(pixels)
        y_test.append(row['emotion'])

# Ép kiểu Numpy Array và Chuẩn hóa dữ liệu (0 -> 1)
X_train = np.array(X_train) / 255.0
X_test = np.array(X_test) / 255.0

# Thêm chiều kênh màu (thành 48x48x1)
X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)

# Mã hóa One-hot cho nhãn (7 cảm xúc)
y_train = to_categorical(np.array(y_train), num_classes=7)
y_test = to_categorical(np.array(y_test), num_classes=7)

print(f"-> Dữ liệu Train: {X_train.shape}, Test: {X_test.shape}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout,BatchNormalization, Activation

```

```python
#Khoi tao mo hinh
model = Sequential()
```

```python
# Lop thu nhat : Tim cac duong net co ban
model.add(Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(48, 48, 1)))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size = (2,2)))
model.add(Dropout(0.25))
```

```python
# Lop thu hai:Tim kiem cac dac diem phuc tap hon
model.add(Conv2D(64, kernel_size=(3, 3), activation='relu'))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size = (2,2)))
model.add(Dropout(0.25))

```

```python
# Lop thu 3: Chuyen Sau
model.add(Conv2D(128, kernel_size=(3, 3), activation='relu'))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D(pool_size = (2,2)))
model.add(Dropout(0.25))

```

```python
model.add(Flatten()) # Duỗi
```

```python
# Tư duy phán đoán
model.add(Dense(256, activation='relu'))
model.add(Dropout(0.5))
```

```python
# Lớp đầu ra
model.add(Dense(7, activation='softmax'))
model.summary()
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Khởi tạo bộ sinh dữ liệu ảnh
aug = ImageDataGenerator(
    rotation_range=15,
    zoom_range=0.15,
    width_shift_range=0.15,
    height_shift_range=0.15,
    horizontal_flip=True
)

# Khởi tạo các tham số huấn luyện
epochs = 50
batch_size = 64

# 2. Biên dịch mô hình
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# 3. Bắt đầu huấn luyện
print("Bắt đầu huấn luyện với Data Augmentation...")
H = model.fit(
    aug.flow(X_train, y_train, batch_size=batch_size),
    validation_data=(X_test, y_test),
    epochs=epochs,
    verbose=1
)
```

```python
model.save('emotion_model.h5')
print("Hoàn tất! Đã lưu file 'emotion_model.h5' thành công.")
```
